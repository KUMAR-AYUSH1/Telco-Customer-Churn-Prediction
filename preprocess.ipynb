feature selection with xgb

```python

cat = df.select_dtypes(include=[ 'category']).columns
print(cat)
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import OrdinalEncoder
from sklearn.compose import ColumnTransformer
col = ['gender', 'InternetService', 'Contract', 'PaymentMethod']
from xgboost import XGBClassifier
x = df.drop(['Churn','customerID'], axis=1)
y = df['Churn']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

ct = ColumnTransformer([('ohe', OrdinalEncoder(), col)], remainder='passthrough')
x_train = ct.fit_transform(x_train)
x_test = ct.transform(x_test)
```

```python
par = {
    'learning_rate': [0.1, 0.01], 
    'max_depth': [3, 4, 5],
    'n_estimators': [100, 150, 200],
    'min_child_weight': [1,5],
    'subsample': [0.8, 1.0]
}

xgb = XGBClassifier(eval_metric='logloss')

grid = GridSearchCV(xgb, par, cv=5, n_jobs=-1, verbose=2)
grid.fit(x_train, y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
y_pred = grid.predict(x_test)
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
model = grid.best_estimator_
feat_importances = pd.Series(model.feature_importances_, index=x.columns).sort_values(ascending=False)

print(feat_importances)
```

```python
top = feat_importances>0.02
top_feat = feat_importances[top]
print(top_feat)
```

```python
df1 = df[top_feat.index]
df  = pd.concat([df1, df['Churn']], axis=1)
df.head()
```

under sampling imbalance data

```python
from imblearn.under_sampling import RandomUnderSampler
rus = RandomUnderSampler(random_state=42)
x_resampled, y_resampled = rus.fit_resample(df.drop('Churn', axis=1), df['Churn'])
df = pd.concat([x_resampled, y_resampled], axis=1)
df['Churn'].value_counts()
```
