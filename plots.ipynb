```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("preprocess.csv")
df.head()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='tenure', hue='Churn', element="step")

plt.title('tenure vs Churned')
plt.show()
```

this shows as tenure increse user less likly to churn , most user churn when they are new
